# Tutorial: Using the `tidy` Command

The `tidy` command in the Aegis CLI is used to clean up and restructure poorly formatted GFF files. This tutorial demonstrates how to use the `tidy` command with an example file.

## Example: Cleaning a Badly Structured GFF

### Input File
We will use the `overlapping_exons.gff3` file located in the `tests/test_data` directory as the input. This file contains overlapping exon features that need to be resolved.

### Command
Run the following command in your terminal:

```bash
aegis tidy \
    tests/test_data/overlapping_exons.gff3 \
    --annotation-name overlapping_exons \
    --output-dir ./tidy_output \
    --output-file overlapping_exons_tidy.gff3 \
    --main \
    --include-UTRs
```

### Explanation of Options
- `--annotation-name`: Specifies the name or tag for the annotation.
- `--output-dir`: Directory where the cleaned GFF file will be saved.
- `--output-file`: Name of the output file.
- `--main`: Includes only the main transcript and CDS per gene.
- `--include-UTRs`: Ensures UTR features are included in the output.

### Output
The cleaned GFF file will be saved in the `tidy_output` directory with the name `overlapping_exons_tidy.gff3`. Open the file to verify that overlapping exons have been resolved and the structure is consistent.

# Tutorial: Using the `prune` Command

The `prune` command in the Aegis CLI is used to remove specific features (e.g., genes or transcripts) from an annotation file. This tutorial demonstrates how to use the `prune` command with an example file.

## Example: Removing Transcripts

### Input Files
- Annotation file: `overlapping_exons.gff3` located in the `tests/test_data` directory.
- ID file: `transcripts_to_remove.txt` containing a list of transcript IDs to be removed.

### Command
Run the following command in your terminal:

```bash
aegis prune \
    tests/test_data/convert_basic.gtf \
    tests/test_data/transcripts_to_remove.txt \
    --annotation-name prune \
    --feature-type transcript \
    --output-dir ./prune_output \
    --output-file convert_basic_pruned
```

### Explanation of Options
- `--annotation-name`: Specifies the name or tag for the annotation.
- `--feature-type`: Indicates the feature level to be removed (e.g., `gene` or `transcript`).
- `--output-dir`: Directory where the pruned file will be saved.
- `--output-file`: Name of the output file.

### Output
The pruned GFF file will be saved in the `prune_output` directory with the name `overlapping_exons_pruned.gff3`. Open the file to verify that the specified transcripts have been removed, and any associated genes have been updated accordingly.

# Tutorial: Using the `rename` Command

The `rename` command in the Aegis CLI is used to automatically rename feature IDs in an annotation file to ensure consistency. This tutorial demonstrates how to use the `rename` command with an example file.

## Example: Renaming Subfeature IDs

### Input File
We will use the `overlapping_exons.gff3` file located in the `tests/test_data` directory as the input.

### Command
Run the following command in your terminal:

```bash
aegis rename \
    tests/test_data/overlapping_exons.gff3 \
    --annotation-name overlapping_exons \
    --output-dir ./rename_output \
    --output-file overlapping_exons_renamed.gff3 \
    --features transcript,CDS,exon,UTR \
    --keep-numbering
```

### Explanation of Options
- `--annotation-name`: Specifies the name or tag for the annotation.
- `--output-dir`: Directory where the renamed file will be saved.
- `--output-file`: Name of the output file.
- `--features`: Specifies which feature levels (e.g., `transcript`, `CDS`, `exon`, `UTR`) will have their IDs renamed.
- `--keep-numbering`: Retains the original numbering of subfeature IDs.

### Output
The renamed GFF file will be saved in the `rename_output` directory with the name `overlapping_exons_renamed.gff3`. Open the file to verify that the IDs have been updated consistently.

# Tutorial: Using the `subset` Command

The `subset` command in the Aegis CLI is used to create a smaller subset of an annotation file based on specific criteria. This tutorial demonstrates how to use the `subset` command to extract only ncRNA features.

## Example: Creating an ncRNA-Only GFF

### Input Files
- Annotation file: `overlapping_exons.gff3` located in the `tests/test_data` directory.
- Genome file: `genome.fasta` (optional, if genome sequences are needed).

### Command
Run the following command in your terminal:

```bash
aegis subset \
    tests/test_data/overlapping_exons.gff3 \
    --annotation-name overlapping_exons \
    --output-dir ./subset_output \
    --gene-cap 0 \
    --chromosomes chr1 \
    --min-genes 0
```

### Explanation of Options
- `--annotation-name`: Specifies the name or tag for the annotation.
- `--output-dir`: Directory where the subset file will be saved.
- `--output-file`: Name of the output file.
- `--gene-cap`: Limits the total number of genes in the subset (set to `0` for no limit).
- `--chromosomes`: Specifies which chromosomes to include in the subset.
- `--min-genes`: Ensures a minimum number of genes in the subset.

### Output
The ncRNA-only GFF file will be saved in the `subset_output` directory with the name `overlapping_exons_ncRNA.gff3`. Open the file to verify that only ncRNA features are included.

# Tutorial: Using the `extract` Command

The `extract` command in the Aegis CLI is used to extract specific features (e.g., proteins, transcripts) from an annotation file. This tutorial demonstrates how to use the `extract` command with an example file.

## Example: Extracting Proteins

### Input Files
- Annotation file: `overlapping_exons.gff3` located in the `tests/test_data` directory.
- Genome file: `genome.fasta` (required for extracting protein sequences).

### Command
Run the following command in your terminal:

```bash
aegis extract \
    tests/test_data/overlapping_exons.gff3 \
    tests/test_data/minimal.fasta \
    --genome-name genome \
    --annotation-name overlapping_exons \
    --output-dir ./extract_output \
    --features protein \
    --mode unique
```

### Explanation of Options
- `--genome-name`: Specifies the name or tag for the genome assembly.
- `--annotation-name`: Specifies the name or tag for the annotation.
- `--output-dir`: Directory where the extracted file will be saved.
- `--output-file`: Name of the output file.
- `--feature-type`: Specifies the feature type to extract (e.g., `protein`).
- `--mode`: Extraction mode (e.g., `unique` to extract unique sequences).

### Output
The extracted protein sequences will be saved in the `extract_output` directory with the name `overlapping_exons_proteins.fasta`. Open the file to verify the extracted sequences.

# Tutorial: Using the `overlaps` Command

The `overlaps` command in the Aegis CLI is used to identify overlapping features within or between annotation files. This tutorial demonstrates how to use the `overlaps` command with an example file.

## Example: Testing Self-Overlaps

### Input File
We will use the `overlapping_exons.gff3` file located in the `tests/test_data` directory as the input.

### Command
Run the following command in your terminal:

```bash
aegis overlap \
    tests/test_data/overlapping_exons.gff3 \
    --annotation-names overlapping_exons \
    --output-dir ./overlap_output \
    --output-filetag overlapping_exons_overlap \
    --overlap-threshold 6 \
    --include-NAs
```

### Explanation of Options
- `--annotation-names`: Specifies the name or tag for the annotation.
- `--output-dir`: Directory where the overlap results will be saved.
- `--output-filetag`: Prefix for the output file name.
- `--overlap-threshold`: Sets the threshold for reporting overlaps.
- `--include-NAs`: Includes non-overlapping features in the output.

### Output
The overlap results will be saved in the `overlap_output` directory with the name `overlapping_exons_overlap`. Open the file to review the reported overlaps.

# Tutorial: Using the `symbols` Command

The `symbols` command in the Aegis CLI is used to add gene symbols to an annotation file based on a tabular import. This tutorial demonstrates how to use the `symbols` command with an example file.

## Example: Adding Gene Symbols

### Input Files
- Annotation file: `overlapping_exons.gff3` located in the `tests/test_data` directory.
- Symbols file: `gene_symbols.tsv` containing tab-separated values of `gene-id` and `gene-symbol`.

### Command
Run the following command in your terminal:

```bash
aegis symbols \
    tests/test_data/overlapping_exons.gff3 \
    tests/test_data/gene_symbols.tsv \
    --annotation-name overlapping_exons \
    --output-dir ./symbols_output \
    --output-file overlapping_exons_with_symbols.gff3 \
    --separator "\t" \
    --header
```

### Explanation of Options
- `--annotation-name`: Specifies the name or tag for the annotation.
- `--output-dir`: Directory where the updated file will be saved.
- `--output-file`: Name of the output file.
- `--separator`: Specifies the separator used in the symbols file (e.g., tab `\t`).
- `--header`: Indicates that the symbols file contains a header row.

### Output
The updated GFF file with gene symbols will be saved in the `symbols_output` directory with the name `overlapping_exons_with_symbols.gff3`. Open the file to verify that the gene symbols have been added.
